```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = pd.read_csv("Unemployment in India.csv")
df2 = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")
```

```python
print("===== DATASET 1 (Unemployment in India) =====")
print(df1.head())
print("\nShape:", df1.shape)

print("\n===== DATASET 2 (Unemployment Rate upto 2020) =====")
print(df2.head())
print("\nShape:", df2.shape)


print("\nColumns in Dataset 1:", df1.columns)
print("Columns in Dataset 2:", df2.columns)
```

```python
print(df1.head())
print(df2.head())


df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()


df1['Date'] = pd.to_datetime(df1['Date'], errors='coerce')
df2['Date'] = pd.to_datetime(df2['Date'], errors='coerce')


df1 = df1.dropna()
df2 = df2.dropna()


print(df1.info())
print(df2.info())
```

```python
df1 = df1.drop_duplicates()
df2 = df2.drop_duplicates()
```

```python
print("\n===== CLEANED DATASET 1 INFO =====")
print(df1.info())

print("\n===== CLEANED DATASET 2 INFO =====")
print(df2.info())
```

```python
print("\nCleaned Dataset 1 Preview:")
print(df1.head())

print("\nCleaned Dataset 2 Preview:")
print(df2.head())
```

```python

print(df1.describe())
print(df2.describe())


print(df1['Region'].unique())
```

```python
plt.figure(figsize=(10,5))
plt.plot(df1['Date'], df1['Estimated Unemployment Rate (%)'])
plt.title("Unemployment Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.lineplot(data=df1, x='Date', y='Estimated Unemployment Rate (%)', hue='Region')
plt.title("Region-wise Unemployment Trends")
plt.show()
```

```python
before_covid = df1[df1['Date'] < '2020-03-01']
after_covid = df1[df1['Date'] >= '2020-03-01']

print("Before Covid Avg:", before_covid['Estimated Unemployment Rate (%)'].mean())
print("After Covid Avg:", after_covid['Estimated Unemployment Rate (%)'].mean())
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=(df1['Date'] >= '2020-03-01'), y='Estimated Unemployment Rate (%)', data=df1)
plt.xticks([0,1], ['Before Covid', 'After Covid'])
plt.title("Covid Impact on Unemployment")
plt.show()
```

```python
df1['Month'] = df1['Date'].dt.month

monthly_avg = df1.groupby('Month')['Estimated Unemployment Rate (%)'].mean()

plt.figure(figsize=(8,5))
monthly_avg.plot(kind='bar')
plt.title("Monthly Unemployment Trend")
plt.xlabel("Month")
plt.ylabel("Avg Unemployment Rate")
plt.show()
```

```python
state_avg = df1.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values()

plt.figure(figsize=(10,6))
state_avg.plot(kind='barh')
plt.title("Average Unemployment by State")
plt.show()
```
